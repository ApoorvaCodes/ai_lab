```python
def unify(x, y):
  if x == y:
    return {}

  if x.islower():
    if x in y:
      return "FAILURE"
      return {x: y}

  if y.islower():
    if y in x:
      return "FAILURE"
      return {y: x}
      return "FAILURE"



x = input("Enter first expression: ")
y = input("Enter second expression: ")

result = unify(x, y)

if result != "FAILURE":
  print("Unification Successful!")
  print("Substitution:", result)
else:
  print("Unification Failed!")
```
