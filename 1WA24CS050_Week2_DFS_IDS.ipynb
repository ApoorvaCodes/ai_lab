```python
def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def dfs(initial, goal):
    stack = [(initial, 0)]
    visited = set()

    while stack:
        state, steps = stack.pop()
        key = tuple(state)

        if key in visited:
            continue

        visited.add(key)

        if state == goal:
            print("Goal state reached!")
            print("Total steps:", steps)
            return

        blank = state.index(0)
        moves = [-3, 3, -1, 1]

        for move in moves:
            new_pos = blank + move

            if 0 <= new_pos < 9:
                if abs(move) == 1 and blank // 3 != new_pos // 3:
                    continue

                new_state = state.copy()
                new_state[blank], new_state[new_pos] = (
                    new_state[new_pos], new_state[blank]
                )

                if tuple(new_state) not in visited:
                    stack.append((new_state, steps + 1))

    print("No solution found.")


initial = [1, 2, 3,
           5, 0, 6,
           4, 7, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

print("Initial State:")
print_board(initial)

print("Goal State:")
print_board(goal)

dfs(initial, goal)
```

```python

def dls(state, goal, depth):
    if state == goal:
        return True

    if depth == 0:
        return False

    z = state.index(0)

    for d in [-3, 3, -1, 1]:
        p = z + d

        if 0 <= p < 9 and not (d == -1 and z % 3 == 0) and not (d == 1 and z % 3 == 2):
            new = state.copy()
            new[z], new[p] = new[p], new[z]

            if dls(new, goal, depth - 1):
                return True

    return False


def iddfs(start, goal, max_depth):
    for depth in range(max_depth + 1):
        if dls(start, goal, depth):
            print("Initial State:")
            for i in range(0, 9, 3):
                print(start[i:i+3])

            print("\nGoal State:")
            for i in range(0, 9, 3):
                print(goal[i:i+3])

            print("\nNumber of Moves:", depth)
            return

    print("Goal not found")


start = [1, 2, 3,
         5, 0, 6,
         4, 7, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

iddfs(start, goal, 10)
```
