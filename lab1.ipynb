```python
grid = [" " for i in range(9)]

def display():
    print(f"{grid[0]} | {grid[1]} | {grid[2]}")
    print("--+---+--")
    print(f"{grid[3]} | {grid[4]} | {grid[5]}")
    print("--+---+--")
    print(f"{grid[6]} | {grid[7]} | {grid[8]}")

def check(player):
    pos = [[0,1,2],[3,4,5],[6,7,8],
           [0,3,6],[1,4,7],[2,5,8],
           [0,4,8],[2,4,6]]
    for p in pos:
        if grid[p[0]] == grid[p[1]] == grid[p[2]] == player:
            return True
    return False

player = "X"
while True:
    display()
    pos = int(input("Enter position (0-8): "))
    if grid[pos] == " ":
        grid[pos] = player
    else:
        print("Invalid move, try again.")
        continue

    if check(player):
        display()
        print("Player", player, "wins!")
        break

    player = "O" if player == "X" else "X"
```

```python


room_a = input("Enter Room A status (Clean/Dirty): ")
room_b = input("Enter Room B status (Clean/Dirty): ")
location = input("Enter vacuum location (A/B): ")

if location == "A":
    if room_a == "Dirty":
        print("Sucking dirt from Room A")
        room_a = "Clean"

    if room_b == "Dirty":
        print("Moving to Room B")
        print("Picking dirt from Room B")
        room_b = "Clean"

elif location == "B":
    if room_b == "Dirty":
        print("Picking dirt from Room B")
        room_b = "Clean"

    if room_a == "Dirty":
        print("Moving to Room A")
        print("Picking dirt from Room A")
        room_a = "Clean"

print("\nFinal State:")
print("Room A:", room_a)
print("Room B:", room_b)

if room_a == "Clean" and room_b == "Clean":
    print("Both rooms are clean!")
```
