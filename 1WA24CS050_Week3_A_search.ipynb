```python

import heapq

def heuristic(state, goal):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1
    return count

def astar(start, goal):
    queue = [(heuristic(start, goal), 0, start)]
    visited = {}

    while queue:
        queue.sort()
        f, moves, state = queue.pop(0)

        if state == goal:
            print("Initial State:")
            for i in range(0, 9, 3):
                print(start[i:i+3])

            print("Goal State:")
            for i in range(0, 9, 3):
                print(goal[i:i+3])

            print("Number of Moves:", moves)
            return

        if tuple(state) in visited and visited[tuple(state)] <= moves:
            continue
        visited[tuple(state)] = moves

        zero = state.index(0)

        for move in [-3, 3, -1, 1]:
            new_pos = zero + move

            if 0 <= new_pos < 9 and not (move == -1 and zero % 3 == 0) and not (move == 1 and zero % 3 == 2):
                new_state = state.copy()
                new_state[zero], new_state[new_pos] = new_state[new_pos], new_state[zero]

                queue.append((moves + 1 + heuristic(new_state, goal),
                              moves + 1, new_state))

start = [2, 8, 3,
         1, 6, 4,
         7, 0, 5]

goal = [1, 2, 3,
        8, 0, 4,
        7, 6, 5]

astar(start, goal)


```

```python

import heapq

def heuristic(state, goal):
    distance = 0

    for i in range(9):
        if state[i] != 0:
            j = goal.index(state[i])
            distance += abs(i // 3 - j // 3) + abs(i % 3 - j % 3)

    return distance

def astar(start, goal):
    queue = [(heuristic(start, goal), 0, start)]
    visited = {}

    while queue:
        queue.sort()
        f, moves, state = queue.pop(0)

        if state == goal:
            print("Initial State:")
            for i in range(0, 9, 3):
                print(start[i:i+3])

            print("Goal State:")
            for i in range(0, 9, 3):
                print(goal[i:i+3])

            print("Number of Moves:", moves)
            return

        if tuple(state) in visited and visited[tuple(state)] <= moves:
            continue
        visited[tuple(state)] = moves

        zero = state.index(0)

        for move in [-3, 3, -1, 1]:
            new_pos = zero + move

            if 0 <= new_pos < 9 and not (move == -1 and zero % 3 == 0) and not (move == 1 and zero % 3 == 2):
                new_state = state.copy()
                new_state[zero], new_state[new_pos] = new_state[new_pos], new_state[zero]

                queue.append((moves + 1 + heuristic(new_state, goal),
                              moves + 1, new_state))

start = [2, 8, 3,
         1, 6, 4,
         7, 0, 5]

goal = [1, 2, 3,
        8, 0, 4,
        7, 6, 5]

astar(start, goal)
```
