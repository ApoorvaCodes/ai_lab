```python
import random

def cost(a):
    c = 0
    for i in range(4):
        for j in range(i+1, 4):
            if abs(a[i]-a[j]) == abs(i-j):
                c += 1
    return c

def hill_climbing():
    a = [3, 1, 2, 0]

    while True:
        best = a[:]
        best_cost = cost(a)

        for i in range(4):
            for j in range(i+1, 4):
                b = a[:]
                b[i], b[j] = b[j], b[i]

                if cost(b) < best_cost:
                    best = b
                    best_cost = cost(b)

        if best == a:
            break

        a = best

    print("Final state:", a)
    print("Cost:", cost(a))

hill_climbing()
```
