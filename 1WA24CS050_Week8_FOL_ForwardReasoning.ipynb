```python
KB = [
    ("American", "Robert"),
    ("Weapon", "T1"),
    ("Sells", "Robert", "T1", "A"),
    ("Hostile", "A")
]
def forward_chaining(KB):

    new_fact = True

    while new_fact:
        new_fact = False

        for a in KB:
            for w in KB:
                for s in KB:
                    for h in KB:

                        if (a[0] == "American" and
                            w[0] == "Weapon" and
                            s[0] == "Sells" and
                            h[0] == "Hostile"):

                            if (a[1] == s[1] and
                                w[1] == s[2] and
                                h[1] == s[3]):

                                fact = ("Criminal", a[1])

                                if fact not in KB:
                                    KB.append(fact)
                                    new_fact = True

    return KB

result = forward_chaining(KB)

print("Knowledge Base after Forward Chaining:")
for fact in result:
    print(fact)

# Check query
if ("Criminal", "Robert") in result:
    print("\nRobert is a Criminal")
else:
    print("\nRobert is not a Criminal")
```
