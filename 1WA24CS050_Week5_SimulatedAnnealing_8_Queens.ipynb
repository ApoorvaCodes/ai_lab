```python
import random
import math

def cost(a):
    c = 0
    for i in range(8):
        for j in range(i+1, 8):
            if a[i] == a[j] or abs(a[i]-a[j]) == abs(i-j):
                c += 1
    return c

def simulated_annealing():
    a = [random.randint(0, 7) for i in range(8)]
    T = 100

    while T > 0.01:
        b = a[:]

        i = random.randint(0, 7)
        b[i] = random.randint(0, 7)

        dE = cost(a) - cost(b)

        if dE > 0:
            a = b
        elif random.random() < math.exp(dE / T):
            a = b

        if cost(a) == 0:
            break

        T = T * 0.99

    print("Final state:", a)
    print("Cost:", cost(a))

simulated_annealing()
```
